# Gateway Aggregation Infrastructure Notebook

This notebook is a companion scaffold for the article. It loads gateway events and supports exploration of child-device freshness, protocol errors, buffer backlog, replay lag, lineage completeness, selective uplink, and site-state quality.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
events_path = root / 'outputs' / 'python_gateway_events.csv'
if events_path.exists():
    events = pd.read_csv(events_path)
else:
    events = pd.read_csv(root / 'data' / 'sample_gateway_events.csv')
events.head()
```

```python
summary = {
    'events': len(events),
    'gateways': events['gateway_id'].nunique(),
    'child_devices': events['device_id'].nunique(),
    'protocol_error_rate': events['protocol_error'].mean(),
    'missing_child_rate': (events['child_device_status'] == 'missing').mean(),
    'mean_buffer_backlog': events['buffer_backlog'].mean(),
    'max_replay_lag_s': events['replay_lag_s'].max(),
    'lineage_completeness_rate': events['lineage_complete'].mean(),
    'selective_uplink_rate': events['forwarded_upstream'].mean()
}
summary
```

# Python Notebook: Edge Interoperability Governance Scoring

This notebook demonstrates how device metadata can be converted into an interoperability governance score.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
data = pd.read_csv(root / 'data' / 'sample_edge_assets.csv')
data.head()
```

```python
data['governance_score'] = (
    0.20 * data['protocol_conformance'] +
    0.20 * data['semantic_alignment'] +
    0.20 * data['lifecycle_control'] +
    0.20 * data['security_baseline'] +
    0.15 * data['operational_accountability'] -
    0.15 * data['unmanaged_divergence']
)
data[['device_id', 'site', 'standard_profile', 'support_state', 'governance_score']]
```

# Embedded Control Notebook: Timing, Saturation, and Safety Filtering

This notebook is a companion scaffold for the article. It loads embedded-control simulation output and supports exploration of tracking error, safety-filter activity, deadline slack, saturation, and supervisory state.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
events_path = root / 'outputs' / 'python_embedded_control_simulation.csv'
if events_path.exists():
    events = pd.read_csv(events_path)
else:
    events = pd.read_csv(root / 'data' / 'sample_control_loop_log.csv')
events.head()
```

```python
summary = {
    'samples': len(events),
    'mean_abs_error': events['control_error'].abs().mean(),
    'saturation_rate': events['saturated'].mean(),
    'deadline_miss_rate': events['deadline_missed'].mean(),
    'min_deadline_slack_ms': events['deadline_slack_ms'].min()
}
summary
```

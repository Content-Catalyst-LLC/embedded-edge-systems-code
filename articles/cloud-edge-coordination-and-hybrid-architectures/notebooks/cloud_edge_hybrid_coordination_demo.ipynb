# Cloud-Edge Hybrid Coordination Notebook

This notebook is a companion scaffold for the article. It loads hybrid-cloud/edge event output and supports exploration of state age, sync lag, policy drift, model skew, rollout convergence, buffer backlog, degraded modes, and authority-window behavior.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
events_path = root / 'outputs' / 'python_cloud_edge_hybrid_events.csv'
if events_path.exists():
    events = pd.read_csv(events_path)
else:
    events = pd.read_csv(root / 'data' / 'sample_hybrid_events.csv')
events.head()
```

```python
summary = {
    'events': len(events),
    'cloud_unreachable_rate': (~events['cloud_reachable']).mean(),
    'degraded_mode_rate': events['degraded_mode'].mean(),
    'policy_drift_rate': (events['edge_policy_version'] != events['cloud_policy_version']).mean(),
    'model_skew_rate': (events['edge_model_version'] != events['approved_model_version']).mean(),
    'rollout_gap_rate': (events['active_version'] != events['target_version']).mean(),
    'authority_violation_rate': (~events['authority_valid']).mean()
}
summary
```

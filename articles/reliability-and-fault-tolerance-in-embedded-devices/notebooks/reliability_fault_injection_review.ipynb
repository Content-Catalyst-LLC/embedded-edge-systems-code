# Reliability Fault Injection Review

Load fault events and summarize detection, recovery, and service-loss evidence.

```python
import pandas as pd
faults = pd.read_csv('../data/fault_events.csv')
faults.groupby('fault_class')[['detected','recovery_success','service_loss_s']].agg({'detected':'mean','recovery_success':'mean','service_loss_s':'sum'})
```

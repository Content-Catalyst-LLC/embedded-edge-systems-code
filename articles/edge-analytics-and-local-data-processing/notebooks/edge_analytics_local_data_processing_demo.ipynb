# Edge Analytics and Local Data Processing Notebook

This notebook is a companion scaffold for the article. It loads analytics events and supports exploration of freshness, local latency, feature completeness, buffer backlog, replay lag, lineage completeness, compression ratio, and uplink modes.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
events_path = root / 'outputs' / 'python_edge_analytics_events.csv'
if events_path.exists():
    events = pd.read_csv(events_path)
else:
    events = pd.read_csv(root / 'data' / 'sample_analytics_events.csv')
events.head()
```

```python
summary = {
    'events': len(events),
    'event_rate': events['event_detected'].mean(),
    'stale_output_rate': (events['freshness_s'] > events['freshness_threshold_s']).mean(),
    'feature_completeness_rate': events['feature_complete'].mean(),
    'lineage_completeness_rate': events['lineage_complete'].mean(),
    'mean_buffer_backlog': events['buffer_backlog'].mean(),
    'mean_replay_lag_s': events['replay_lag_s'].mean(),
    'compression_ratio': 1 - events['uplink_bytes'].sum() / events['raw_bytes'].sum()
}
summary
```

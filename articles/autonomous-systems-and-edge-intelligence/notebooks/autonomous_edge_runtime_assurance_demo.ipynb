# Autonomous Edge Notebook: Belief, Runtime Assurance, and Drift

This notebook is a companion scaffold for the article. It loads the autonomy simulation output and supports exploration of belief states, candidate actions, safety-filtered actions, fallback behavior, and drift monitoring.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
events_path = root / 'outputs' / 'python_autonomous_edge_decision_events.csv'
if events_path.exists():
    events = pd.read_csv(events_path)
else:
    events = pd.read_csv(root / 'data' / 'sample_autonomy_events.csv')
events.head()
```

```python
summary = {
    'decisions': len(events),
    'fallback_rate': (events.get('action_type', pd.Series(dtype=str)) == 'fallback').mean(),
    'mean_confidence': events['decision_confidence'].mean(),
    'latency_violation_rate': (events['latency_ms'] > events['latency_budget_ms']).mean()
}
summary
```

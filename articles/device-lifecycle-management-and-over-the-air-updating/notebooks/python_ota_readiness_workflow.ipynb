# Python Notebook: OTA Fleet Readiness and Rollout Risk Scoring

This notebook demonstrates how lifecycle metadata can be converted into OTA readiness and rollout decisions.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
fleet = pd.read_csv(root / 'data' / 'sample_device_fleet.csv')
fleet.head()
```

```python
fleet['ota_readiness_score'] = (
    0.16 * fleet['identity_assurance'] +
    0.18 * fleet['compatibility_match'] +
    0.16 * fleet['package_integrity'] +
    0.16 * fleet['validation_status'] +
    0.16 * fleet['rollback_readiness'] +
    0.13 * fleet['observability'] -
    0.15 * fleet['lifecycle_drift']
)
fleet[['device_id', 'site', 'support_state', 'rollout_ring', 'ota_readiness_score']]
```

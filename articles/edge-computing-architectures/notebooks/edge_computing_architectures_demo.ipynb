Explore workload placement, latency, offline readiness, version skew, trust, runtime assurance, and rollback readiness.

```python
from pathlib import Path
import pandas as pd
root = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
fleet = pd.read_csv(root / 'data' / 'sample_edge_fleet_inventory.csv')
fleet.head()
```

```python
{
    'assets': len(fleet),
    'online_rate': (fleet['connectivity_state'] == 'online').mean(),
    'version_skew_rate': (fleet['active_version'] != fleet['approved_version']).mean(),
    'trust_verified_rate': (fleet['trust_state'] == 'verified').mean(),
    'runtime_ready_rate': (fleet['runtime_assurance_state'] == 'ready').mean(),
    'rollback_ready_rate': fleet['rollback_ready'].mean()
}
```

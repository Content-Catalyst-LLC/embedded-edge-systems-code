# Low-Power Budget Review

Load state-level power data and estimate energy share by operating state.

```python
import pandas as pd
states = pd.read_csv('../data/power_states.csv')
states['energy_mj'] = states['current_ma'] * states['voltage_v'] * states['duration_s']
states[['state_name','current_ma','duration_s','energy_mj']].sort_values('energy_mj', ascending=False)
```
